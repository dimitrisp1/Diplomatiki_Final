# UMAP 2D Projection Figure
Fresh 2-component UMAP fit (same n_neighbors/min_dist as the selected
solution) for visualisation only; the clustering itself used 13 components.

```python
import pandas as pd, numpy as np, warnings, umap
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize
from sklearn.decomposition import PCA
warnings.filterwarnings("ignore")

df = pd.read_csv("tovima_final.csv", encoding="utf-32", sep="\t")
emb = normalize(np.load("tovima_embeddings_bge_m3.npy"), norm="l2")
pca = PCA(n_components=256, svd_solver="randomized", random_state=42)
X = pca.fit_transform(emb)
```

```python
xy = umap.UMAP(n_neighbors=11, n_components=2, min_dist=0.0141,
               metric="cosine", random_state=42).fit_transform(X)
```

```python
topics = [t for t in df["major_topic"].value_counts().index if t != "Noise"]
cmap = plt.cm.tab20(np.linspace(0, 1, len(topics)))
fig, ax = plt.subplots(figsize=(10, 7.5), dpi=110)
nm = df["major_topic"] == "Noise"
ax.scatter(xy[nm,0], xy[nm,1], s=2, c="#cccccc", alpha=.5, label="Noise", rasterized=True)
for t, c in zip(topics, cmap):
    m = (df["major_topic"] == t).values
    ax.scatter(xy[m,0], xy[m,1], s=2.5, color=c, alpha=.7, label=t, rasterized=True)
ax.set_xticks([]); ax.set_yticks([])
ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1), frameon=False, fontsize=8, markerscale=5)
ax.set_title("UMAP 2-D projection of the corpus, coloured by major topic")
plt.tight_layout(); plt.savefig("fig_umap_scatter.png", dpi=200, bbox_inches="tight"); plt.close()
```

```python
sizes = df[df.final_label != -1].final_label.value_counts().values
fig, ax = plt.subplots(figsize=(8, 3.6), dpi=110)
ax.hist(sizes, bins=40, color="#4472a8", edgecolor="white")
ax.set_xlabel("cluster size (emails)"); ax.set_ylabel("clusters")
ax.set_title("Cluster size distribution (107 clusters)")
plt.tight_layout(); plt.savefig("fig_cluster_sizes.png", dpi=200, bbox_inches="tight"); plt.close()
print("figures written")
```
